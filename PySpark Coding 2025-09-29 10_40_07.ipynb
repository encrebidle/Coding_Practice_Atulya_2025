```python
#Volumes/workspace/default/sparkpractice/sample.csv
#/Volumes/workspace/default/sparkpractice/Fire_Department_and_Emergency_Medical_Services_Dispatched_Calls_for_Service_20250926.csv
import pyspark
fire_department_csv = r"/Volumes/workspace/default/sparkpractice/Fire_Department_and_Emergency_Medical_Services_Dispatched_Calls_for_Service_20250926.csv"
```

```python
fire_df = spark.read \
    .format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(fire_department_csv)
display(fire_df)
```

```python
len(fire_df.columns)
```

```python
fire_df.createOrReplaceTempView("fire_service_calls_view")
```

```python
%sql
select * from fire_service_calls_view limit 10
```

```python
%sql
create database if not exists demo_db
```

```python
# Rename columns to valid Delta Lake column names
df = spark.table("fire_service_calls_view")
for col in df.columns:
    new_col = col.replace(" ", "_").replace("-", "_").replace("/", "_")
    df = df.withColumnRenamed(col, new_col)

# Create the table with the cleaned column names
df.write.format("delta").mode("overwrite").saveAsTable("demo_db.fire_service_calls")
display(spark.table("demo_db.fire_service_calls"))
```
