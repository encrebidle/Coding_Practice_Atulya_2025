## Big Data

### 1. Origins and Evolution of Data Processing
- Early data processing began with **programming languages like COBOL**.  
- Progressed to **Relational Database Management Systems (RDBMS)** such as **Oracle**.  
- These systems introduced structured storage, indexing, and SQL queries.  

RDBMS became the foundation for enterprise data management, but with time, new challenges emerged.

---

### 2. Limitations of Traditional RDBMS
- Struggled with the **3 Vs of Big Data**:
  - **Volume** – massive amounts of data generated.  
  - **Variety** – structured, semi-structured (XML, JSON), and unstructured formats (images, videos, logs).  
  - **Velocity** – speed at which data is generated and processed.  
- RDBMS works well for structured data but is inefficient for semi-structured/unstructured data.  

This gap is referred to as the **Big Data problem**.

---

### 3. Approaches to Data Processing
- **Monolithic Approach**  
  - One large system handling everything.  
  - Expensive and hard to scale.  

- **Distributed Approach**  
  - Uses a **cluster of systems** to share workload.  
  - More **scalable, cost-effective, and fault-tolerant**.  

The distributed approach is the foundation of modern Big Data technologies.

---

### 4. Hadoop
- Hadoop was developed to solve Big Data challenges.  
- Works as a **distributed cluster** for:
  - **Storage** – data spread across multiple nodes.  
  - **Retrieval** – data accessed from the cluster efficiently.  
  - **Processing** – computation done using **MapReduce**.  

MapReduce enables parallel processing of large datasets across multiple machines.

---

### 5. Hadoop vs RDBMS
| Feature            | RDBMS (e.g., Oracle)         | Hadoop                       |
|--------------------|------------------------------|------------------------------|
| Data Type          | Structured only              | Structured, semi-structured, unstructured |
| Scalability        | Limited, vertical scaling    | High, horizontal scaling     |
| Cost               | Expensive hardware/software  | Runs on commodity hardware   |
| Processing Model   | Centralized, sequential      | Distributed, parallel (MapReduce) |
| Best For           | Transactions, structured queries | Large-scale storage & processing |

---

### 6. Key Points
- Traditional RDBMS cannot handle Big Data effectively due to the 3 Vs.  
- Distributed systems provide scalability and flexibility.  
- Hadoop emerged as the go-to solution for Big Data with its cluster-based design.  
- Next: deeper study of the Hadoop ecosystem and its tools.


## Hadoop and Its Evolution

### 1. What is Hadoop?
- A **distributed data processing framework** designed to handle Big Data.  
- Breaks data and tasks into smaller parts, spreads them across a **cluster of machines**, and processes them in parallel.  
- Originated from Google’s papers on **GFS (Google File System)** and **MapReduce**.  

Analogy: Like dividing a giant dictionary into many sections and giving them to different people to sort. Together they finish faster than one person.

---

### 2. Core Components of Hadoop
1. **HDFS (Hadoop Distributed File System)** – storage  
   - Splits files into **blocks** (default: 128 MB).  
   - Stores blocks across multiple machines.  
   - **NameNode** (master) keeps track of metadata (which block is where).  
   - **DataNodes** (workers) hold the actual data.  

   Example: A 500 MB file = 4 blocks → stored on different nodes, replicated for fault tolerance.  

2. **YARN (Yet Another Resource Negotiator)** – resource manager  
   - Allocates cluster resources (CPU, memory).  
   - Manages multiple jobs at the same time.  

   Analogy: Like an airport traffic controller assigning runways to different flights.  

3. **MapReduce** – processing model  
   - **Map phase:** divides the task into smaller subtasks.  
   - **Reduce phase:** combines the results into final output.  

   Example: Counting words in 1 million documents.  
   - Map → Each node counts words in its share.  
   - Reduce → Combine all counts for final result.

---

### 3. Hadoop Architecture (Quick Map)

- **HDFS**  
  - NameNode → metadata  
  - DataNodes → store blocks  
- **YARN**  
  - ResourceManager → allocates tasks  
- **MapReduce**  
  - Map → divide task  
  - Reduce → aggregate results  

---

### 4. Strengths of Hadoop
- Handles **very large data** (TB–PB scale).  
- Runs on **commodity hardware** (no need for supercomputers).  
- **Fault-tolerant** → if a node fails, data is available elsewhere.  

Analogy: Like storing the same book in multiple library branches — if one branch burns down, copies exist in others.

---

### 5. Limitations of Hadoop (MapReduce)
- Writing MapReduce code is **complex** (mostly Java).  
- **Slow** – relies on disk for every stage of processing.  
- Suited only for **batch processing** (not real-time).  

Comparison: Like a **diesel train** – strong and reliable but slow and not flexible.

---

### 6. Apache Spark – The Next Step
- Developed to overcome Hadoop’s performance and complexity issues.  
- **In-memory processing** → much faster than MapReduce.  
- Supports multiple languages: **Python, Scala, Java, R**.  
- Works for **batch, streaming, machine learning, and graph** workloads.  
- Can run on top of Hadoop (using HDFS/YARN) or standalone with cloud storage.  

Analogy: If Hadoop (MapReduce) is a **diesel train**, Spark is a **bullet train** – same goal (move data), but much faster, more efficient, and can handle diverse tasks.

---

### 7. Hadoop vs Spark

| Aspect              | Hadoop (MapReduce)              | Apache Spark                     |
|---------------------|---------------------------------|----------------------------------|
| Processing Speed    | Disk-based, slower              | In-memory, much faster           |
| Ease of Use         | Complex (Java)                  | Easy APIs (Python, Scala, etc.)  |
| Workloads           | Batch only                      | Batch + Streaming + ML + Graph   |
| Scalability         | Horizontal, but slower          | Horizontal + in-memory, faster   |
| Deployment          | On-premise clusters             | On-premise + Cloud               |

---

### 8. Key Points
- Hadoop pioneered **distributed Big Data processing**.  
- MapReduce introduced parallel computation but had **speed and usability issues**.  
- Apache Spark is now the **industry standard** due to speed, flexibility, and simplicity.  
- Hadoop’s role today → often used for **storage (HDFS)** and **resource management (YARN)**, while Spark is used for processing.


## Data Lakes

### 1. Evolution from Data Warehouses
- **Data Warehouses (DWs)**  
  - Designed for **structured data** (tables, rows, columns).  
  - Optimized for **reporting, dashboards, and BI**.  
  - Strong in **transaction consistency** (ACID properties).  

- **Data Lakes**  
  - Emerged as an alternative to handle **all types of data**:  
    - Structured (tables),  
    - Semi-structured (JSON, XML, logs),  
    - Unstructured (images, videos, PDFs).  
  - Built on top of **distributed systems** (like HDFS).  

**Analogy:**  
- A Data Warehouse is like a **filtered bottled water plant** – clean, structured, ready to drink.  
- A Data Lake is like a **natural lake** – contains everything: clean water, mud, fish, plants — raw, unfiltered.  

---

### 2. The Role of Distributed Computing
- Before Hadoop, storage and computation were expensive and centralized.  
- **HDFS (Hadoop Distributed File System)** enabled cheap, scalable storage across clusters.  
- **MapReduce** allowed parallel processing of massive data sets.  
- Together, they created the foundation for the **Data Lake model**.  

---

### 3. Origin of the Term "Data Lake"
- Coined by **James Dixon** (then CTO of Pentaho).  
- His idea: instead of multiple structured ponds (Data Marts/Warehouses), store data in a **large reservoir (Lake)** where it remains raw and can be used in different ways.  

---

### 4. Challenges of Early Data Lakes
- **Transaction consistency** (ACID) was weak compared to Data Warehouses.  
- Lacked **governance and access control**.  
- Often turned into **“data swamps”** when not managed well — messy, disorganized, hard to extract value.  

---

### 5. Hybrid Approach – Lakehouse
- New architectures combine the **flexibility of Data Lakes** with the **reliability of Data Warehouses**.  
- Known as **Data Lakehouse** (popularized by Databricks).  
- Benefits:  
  - Store any kind of data (Lake).  
  - Ensure schema, transactions, and performance (Warehouse).  
  - Enable both **analytics (BI)** and **AI/ML workloads**.  

**Example:**  
- A retailer stores raw clickstream logs (Data Lake) + sales transactions (Data Warehouse).  
- Together in a Lakehouse → advanced analytics + accurate reporting.

---

### 6. Cloud Infrastructure for Data Lakes
- Cloud made Data Lakes **cheaper and more scalable**.  
- Examples:  
  - **Amazon S3** – object storage for raw and processed data.  
  - **Azure Blob Storage** – scalable cloud storage for unstructured data.  
  - **Google Cloud Storage (GCS)** – similar scalable storage.  

These provide virtually **infinite storage at low cost**, with pay-as-you-go models.

---

### 7. Key Capabilities of Modern Data Lakes
1. **Data Ingestion** – from multiple sources (databases, IoT devices, logs, APIs).  
2. **Data Management** – partitioning, cataloging, and schema evolution.  
3. **Data Processing** – batch, streaming, machine learning workloads.  
4. **Data Retrieval** – query engines (Presto, Hive, Spark SQL).  
5. **Governance & Security** – access control, encryption, compliance (GDPR, HIPAA).  

---

### 8. Challenges in Implementation
- **Data Governance** – preventing “data swamp” scenario.  
- **Security & Access Control** – who can access what data.  
- **Schema Evolution** – managing constantly changing data formats.  
- **Performance** – querying raw data can be slower than in Warehouses.  

---

### 9. Data Warehouses vs Data Lakes vs Lakehouse

| Feature               | Data Warehouse          | Data Lake                 | Lakehouse (Hybrid)        |
|------------------------|-------------------------|---------------------------|---------------------------|
| Data Type              | Structured only         | All types (raw + unstructured) | All types (with structure when needed) |
| Schema                 | Schema-on-write         | Schema-on-read            | Both (flexible + reliable) |
| Cost                   | Expensive               | Cheaper (commodity/cloud) | Balanced (cloud-optimized) |
| Best For               | BI, reporting           | Raw storage, ML/AI        | BI + ML + real-time apps  |
| Examples               | Oracle, Snowflake       | Hadoop, S3, Blob Storage  | Databricks Delta Lake, Snowflake (modern) |

---

### 10. Key Takeaways
- Data Lakes evolved from Warehouses to handle **volume, variety, velocity** of modern data.  
- HDFS and MapReduce enabled the first affordable, scalable Data Lakes.  
- Early challenges (consistency, governance) led to **Lakehouse architectures**.  
- Cloud (AWS, Azure, GCP) made storage **scalable and low-cost**.  
- Modern Data Lakes are central to **AI, analytics, and enterprise data platforms**.


## Apache Spark and Databricks Cloud

### 1. What is Apache Spark?
- An **open-source distributed computing framework** designed for **fast and scalable data processing**.  
- Works across clusters of machines (horizontal scaling).  
- Can integrate with various **cluster managers** (YARN, Mesos, Kubernetes) and **storage systems** (HDFS, Amazon S3, Azure Blob, GCS).  

**Analogy:** Spark is like the **engine of a factory** — it doesn’t store raw materials itself (that’s HDFS/S3) and doesn’t own the factory building (that’s the cluster manager), but it powers all the production work.

---

### 2. Spark Architecture & Ecosystem
- **Two Layers:**
  1. **Core Layer**  
     - Distributed computing engine.  
     - Handles scheduling, memory, fault tolerance.  
     - Provides the **low-level RDD API**.  
  2. **Libraries, APIs, and DSLs** (Domain-Specific Languages)  
     - Built on top of the core to make development easier.  

- **Core APIs available in:** Scala, Java, Python, R.  
  - Scala = native, fastest.  
  - Python (PySpark) = most popular among data scientists.  

---

### 3. Spark Compute Engine – Responsibilities
- **Task Scheduling** – deciding how tasks are divided and executed on worker nodes.  
- **Data Management** – caching, shuffling, partitioning data for efficiency.  
- **Fault Tolerance** – if a node fails, Spark recomputes lost partitions using lineage.  

**Example:** If a worker crashes while processing, Spark reassigns the same data partition to another worker using the input lineage graph.

---

### 4. APIs and Libraries in Spark
1. **RDD (Resilient Distributed Dataset)** – the original abstraction  
   - Immutable, fault-tolerant, low-level.  
   - Powerful but complex.  

2. **Spark SQL / DataFrames / Datasets**  
   - High-level APIs.  
   - Support both **SQL queries** and **functional programming** (e.g., `map`, `filter`).  
   - Easier to use than RDDs.  

3. **Spark Streaming**  
   - Real-time data processing.  
   - Works with event streams (Kafka, Flume, sockets).  

4. **MLlib**  
   - Built-in machine learning library (classification, regression, clustering).  

5. **GraphX**  
   - Graph analytics library.  

**Analogy:**  
- RDD = raw ingredients (flexible but hard to handle).  
- DataFrame/SQL = ready-to-cook meal kits (simpler, faster to use).  
- Libraries = specialized tools (blender for ML, strainer for streaming, etc.).

---

### 5. Why Spark is Popular
- **Unified framework** → batch + streaming + ML + graph, all in one.  
- **Language flexibility** → use Python, Scala, Java, R.  
- **Speed** → in-memory processing, much faster than Hadoop MapReduce.  
- **Simplifies distributed computing** → hides complexity of cluster operations.

---

### 6. Databricks Cloud
- A **cloud-based platform built around Spark**.  
- Founded by the original creators of Spark.  
- Provides a **managed Spark environment** with additional tools.  

**Key Features**
1. **Cluster Management** – easily launch Spark clusters on AWS, Azure, GCP.  
2. **Collaborative Notebooks** – Python, SQL, R, Scala in one workspace (like Jupyter + GitHub).  
3. **Optimized Spark Runtime** – faster execution, autoscaling clusters.  
4. **Photon Engine** – boosts SQL performance.  
5. **Delta Lake** – provides ACID transactions and schema enforcement in Data Lakes.  
6. **MLflow** – experiment tracking, ML model management.  

**Analogy:** If Apache Spark is a **powerful engine**, then Databricks is a **luxury car built around that engine** with better controls, safety, and features.

---

### 7. Spark vs Databricks

| Aspect                  | Apache Spark                  | Databricks Cloud                     |
|--------------------------|--------------------------------|---------------------------------------|
| Core Role                | Distributed processing engine | Cloud platform built on Spark          |
| Cluster Management       | Needs external (YARN, K8s, Mesos)    | Fully managed (AWS, Azure, GCP)       |
| Usability                | Command-line, code heavy      | GUI, notebooks, collaboration         |
| Performance              | Fast                          | Faster (optimized runtime + Photon)   |
| Advanced Features        | Libraries only                | Delta Lake, MLflow, security, autoscale |

---

### 8. Key Takeaways
- **Spark** = foundation for distributed data processing.  
- **Core engine** handles scheduling, fault tolerance, APIs (RDDs, DataFrames).  
- Spark’s **ecosystem** makes it versatile: SQL, streaming, ML, graph.  
- **Databricks** takes Spark further → easier cluster management, better performance, collaborative tools, and enterprise features.  
- In modern **Data Lakes and Lakehouses**, Spark + Databricks are central to analytics and AI workloads.


## Spark DataFrames

### Spark vs Traditional Databases
- Both use **tables** and **SQL** for data processing.  
- Traditional database has three layers:  
  - **Storage Layer** – actual data.  
  - **Metadata Layer** – schema and table definitions.  
  - **Logical Layer** – queries and SQL execution.  
- Schema is central because SQL queries cannot run without it.  

Spark follows the same logic but executes in a distributed environment.

---

### File Formats and Data Types
- Spark supports multiple formats: **CSV, JSON, Parquet, AVRO**.  
- Can handle **structured**, **semi-structured**, and **unstructured** data.  
- This gives Spark higher scalability than traditional databases.  

---

### Spark DataFrames
- Similar to a database table but with differences:  
  - Schema is not stored in a metadata store.  
  - Uses **schema-on-read**, meaning schema is applied while loading data.  
  - Exists only as a temporary, in-memory object.  
- Processed through **DataFrame APIs**.  

Think of a DataFrame as a session-level table used for transformations and analysis.

---

### Spark Tables vs DataFrames

| Aspect          | Spark Tables                        | Spark DataFrames                 |
|-----------------|-------------------------------------|----------------------------------|
| Schema          | Persistent in metadata store        | Defined during load (schema-on-read) |
| Storage         | Permanent                           | Temporary, in-memory             |
| Usage           | SQL queries                         | DataFrame API operations         |

Tables are long-term storage objects, while DataFrames are short-term working objects.  

---

### Key Points
- Spark extends the database idea but is more flexible and scalable.  
- **Tables**: permanent, schema fixed in metadata.  
- **DataFrames**: temporary, schema applied only when data is read.
